## 📖 Concept: What Is a CSV and How Do We Load It?

### What Is CSV?

**CSV** stands for **Comma-Separated Values**. It is one of the simplest and most common file formats for storing tabular data.

A CSV file looks like this internally:

```
Pos,Name,Value
1,American Samoa,74.60
2,Nauru,71.10
3,Cook Islands,63.70
```

- The **first row** is usually the **header** (column names)
- Each subsequent row is a **record** (one observation/entry)
- Values are separated by **commas**

### `pd.read_csv()`

This is the pandas function that reads a CSV file and converts it into a **DataFrame**.

```python
df = pd.read_csv("path/to/file.csv")
```

| Parameter | What it does |
|-----------|--------------|
| `filepath` | The path to your CSV file (relative or absolute) |
| `sep` | The delimiter character (default is `,`) |
| `header` | Which row to use as column names (default is `0`, the first row) |
| `encoding` | Character encoding (e.g., `'utf-8'`, `'latin-1'`) |

### What Is a DataFrame?

A **DataFrame** is a 2-dimensional labelled data structure — like a spreadsheet or SQL table. It has:
- **Rows** (each with an index label)
- **Columns** (each with a name)
- **Cells** (individual values at the intersection of a row and column)

### Inspecting Your Data

After loading, you should always inspect the data:

| Method | What it shows |

| `df.dtypes` | Data type of each column |
| `df.info()` | Summary: column names, types, non-null counts |
| `df.describe()` | Statistical summary of numeric columns |

# Section 3: Renaming Columns

---

## 📖 Concept: Why and How Do We Rename Columns?

### Why Rename?

Our two datasets both have a column called **"Value"** — but one holds obesity rates and the other holds population counts. If we tried to merge them right now, pandas wouldn't know which "Value" is which. Renaming makes our data **self-documenting** and prevents confusion.

Good column names are:
- **Descriptive** — "Obesity Rate" is better than "Value"
- **Consistent** — use the same naming convention across datasets
- **Merge-ready** — shared columns (like country name) must have the **exact same name** in both DataFrames for merging to work

### The `.rename()` Method

```python
df.rename(columns={"old_name": "new_name"}, inplace=True)
```

| Parameter | What it does |
|-----------|--------------|
| `columns` | A **dictionary** mapping old names → new names |
| `inplace` | If `True`, modifies the DataFrame directly. If `False` (default), returns a **new** DataFrame |

### Understanding `inplace`

This is a critical concept:

```python
# WITHOUT inplace — creates a new DataFrame (original is unchanged)
new_df = df.rename(columns={"A": "B"})

# WITH inplace — modifies the original DataFrame directly
df.rename(columns={"A": "B"}, inplace=True)
```

### Understanding Dictionaries (Quick Refresher)

The `columns` parameter takes a **dictionary** — a collection of key-value pairs:

```python
my_dict = {
    "key1": "value1",
    "key2": "value2"
}
```

In our case, the keys are the **old column names** and the values are the **new column names**.

## 💻 Case Study Code

````python
# Rename columns in sample1 (obesity data)
sample1.rename(
    columns={
        "Pos": "Position",
        "Name": "Country",
        "Value": "Obesity Rate"
    }, inplace=True
)

# Rename columns in sample2 (population data)
sample2.rename(
    columns={
        "Pos": "Position",
        "Name": "Country",
        "Value": "Population"
    }, inplace=True
)
````

**What just happened?**

We renamed all three columns in both DataFrames:

| Original | sample1 (new) | sample2 (new) |
|----------|---------------|---------------|
| `Pos` | `Position` | `Position` |
| `Name` | `Country` | `Country` |
| `Value` | `Obesity Rate` | `Population` |

Now the **"Country"** column has the same name in both datasets (essential for merging), and the value columns have descriptive names.

Let's verify:

````python
print("sample1 columns:", list(sample1.columns))
print("sample2 columns:", list(sample2.columns))
print()
sample1.head(3)
````

# Section 4: Merging (Joining) Two DataFrames

---

## 📖 Concept: What Is a Merge and What Are Join Types?

### The Problem

We have two separate tables with related data. The **obesity** data is in one table and the **population** data is in another. To analyse them together, we need to **combine them into a single table** based on a common column.

This operation is called a **merge** (or **join**) — the same concept as a SQL JOIN.

### `pd.merge()` Syntax

```python
result = pd.merge(left_df, right_df, how="join_type", on="common_column")
```

| Parameter | What it does |
|-----------|--------------|
| `left_df` | The first (left) DataFrame |
| `right_df` | The second (right) DataFrame |
| `how` | The type of join to perform |
| `on` | The column(s) to join on (must exist in both DataFrames) |

### The Four Join Types

Understanding join types is **fundamental** to data management:

| Join Type | What It Keeps | Visual |
|-----------|---------------|--------|
| `"inner"` | Only rows where the key exists in **BOTH** tables | A ∩ B |
| `"left"` | ALL rows from the **left** table + matching rows from right | A + (A ∩ B) |
| `"right"` | ALL rows from the **right** table + matching rows from left | B + (A ∩ B) |
| `"outer"` | ALL rows from **BOTH** tables | A ∪ B |

### Visual Example

Imagine two small tables:

**Table A (Obesity):**
| Country | Obesity Rate |
|---------|--------------|
| USA | 33.0 |
| Nauru | 71.1 |

**Table B (Population):**
| Country | Population |
|---------|-----------|
| USA | 321,368,864 |
| China | 1,367,485,388 |

- **Inner join:** Only USA (exists in both)
- **Left join:** USA + Nauru (all from left; Nauru's Population = NaN)
- **Right join:** USA + China (all from right; China's Obesity Rate = NaN)
- **Outer join:** USA + Nauru + China (all from both; missing values = NaN)

### What Happens to Duplicate Column Names?

If both DataFrames have columns with the same name (other than the join key), pandas appends **suffixes** to distinguish them: `_x` (from left) and `_y` (from right). In our case, both tables have a "Position" column, so the merged result will have `Position_x` and `Position_y`.

## 💻 Case Study Code

````python
# Merge the two datasets on the "Country" column using a left join
joinedTable = pd.merge(sample1, sample2, how="left", on="Country")
````

**What just happened?**

- We merged `sample1` (obesity, 191 countries) with `sample2` (population, 238 countries)
- We used a **left join** (`how="left"`), which means:
  - ✅ Every country in `sample1` is kept
  - ✅ If a matching country exists in `sample2`, its population is added
  - ⚠️ If a country in `sample1` has **no match** in `sample2`, the Population column will show `NaN`
  - ❌ Countries that are only in `sample2` (but not in `sample1`) are **excluded**

Let's inspect the result:

````python
print(f"Merged DataFrame shape: {joinedTable.shape}")
print(f"Columns: {list(joinedTable.columns)}")
print()
joinedTable.head(10)
````

> 🔍 **Notice:** The result has columns `Position_x`, `Country`, `Obesity Rate`, `Position_y`, and `Population`. The "Position" column appeared in both DataFrames, so pandas added `_x` and `_y` suffixes. We'll select only the columns we need in the next step.

After merging, our DataFrame has more columns than we need. In data management, **selecting only the columns you need** is called a **projection**. This:

- Reduces memory usage

# Section 6: Filtering Rows with Boolean Conditions

---

## 📖 Concept: Boolean Indexing and `.loc[]`

### What Is Filtering?

Filtering means keeping only the rows that satisfy a given **condition**. In SQL, this is the `WHERE` clause. In pandas, we use **Boolean indexing**.

### How It Works — Step by Step

**Step 1:** Create a Boolean condition
```python
df["Population"] > 10000000
```
This returns a **Series of True/False values** — one for each row.

**Step 2:** Use that Series to select rows
```python
df[df["Population"] > 10000000]
# or equivalently:
df.loc[df["Population"] > 10000000]
```
Only rows where the condition is `True` are kept.

### Combining Multiple Conditions

Use these operators (NOT the English words `and`, `or`):

| Operator | Meaning | Example |
|----------|---------|---------|
| `&` | AND | `(cond1) & (cond2)` |
| `\|` | OR | `(cond1) \| (cond2)` |
| `~` | NOT | `~(cond1)` |

### ⚠️ Critical Rule: Parentheses Are Required!

When combining conditions, each condition **must** be wrapped in parentheses:

```python
# ✅ CORRECT
df.loc[(df["A"] > 5) & (df["B"] < 10)]

# ❌ WRONG — will cause an error
df.loc[df["A"] > 5 & df["B"] < 10]
```

This is because `&` has **higher precedence** than `>` and `<` in Python.

### `.loc[]` vs `[]`

Both can filter rows, but `.loc[]` is more explicit and supports simultaneous row and column selection:

```python
# Filter rows AND select columns in one step
df.loc[df["A"] > 5, ["A", "B"]]
```

## 💻 Case Study Code

````python
# Filter: Obesity Rate > 20 AND Population > 10,000,000
obesity_population = joinedTable1.loc[
    (joinedTable1["Population"] > 10000000) & (joinedTable1["Obesity Rate"] > 20)
]
````

**What just happened?**

We applied **two filters simultaneously** using the `&` (AND) operator:

1. `joinedTable1["Population"] > 10000000` → keeps only countries with population > 10 million
2. `joinedTable1["Obesity Rate"] > 20` → keeps only countries with obesity rate > 20%

The `&` means **both conditions must be True** for a row to be included.

Let's see how many countries survived the filter:

````python
print(f"Before filtering: {joinedTable1.shape[0]} countries")
print(f"After filtering:  {obesity_population.shape[0]} countries")
print()
obesity_population.head(10)
````

# Section 7: Sorting Data

---

## 📖 Concept: Ordering Your Data with `.sort_values()`

### Why Sort?

Sorting organises your data in a meaningful order — whether you want to find the **highest**, **lowest**, or **alphabetically first** entries. It's one of the most common operations in data analysis.

### `sort_values()` Syntax

```python
df.sort_values(by="column_name", ascending=True/False)
```

| Parameter | What it does | Default |
|-----------|--------------|---------|
| `by` | The column(s) to sort by | (required) |
| `ascending` | `True` = smallest first (A→Z, 1→100). `False` = largest first (Z→A, 100→1) | `True` |
| `inplace` | If `True`, modifies the DataFrame directly | `False` |
| `na_position` | Where to put `NaN` values: `'first'` or `'last'` | `'last'` |

### Sorting by Multiple Columns

You can sort by multiple columns by passing a **list**:

```python
df.sort_values(by=["col1", "col2"], ascending=[False, True])
```

This sorts by `col1` descending first, then by `col2` ascending for ties.

## 💻 Case Study Code

````python
# Sort by Obesity Rate in descending order (highest first)
sorting = obesity_population.sort_values(by="Obesity Rate", ascending=False)
````

# Section 8: Slicing the Top-N and Resetting the Index

---

## 📖 Concept: `.head()` for Top-N and Index Management

### Getting the Top N Rows

After sorting, you often want just the **top N** results. The `.head(n)` method returns the first `n` rows:

```python
df.head(10)  # first 10 rows
```

### Understanding the Index

Every DataFrame has an **index** — a label for each row. By default, it's a sequence of integers starting from 0. But after filtering and sorting, the original index values are **preserved**, which can look messy:

```
     Country  Population  Obesity Rate
16   Egypt    88487396    33.1
17   United States  321368864  33.0
18   Saudi Arabia   27752316   33.0
```

The index values (16, 17, 18) are from the original DataFrame — not ideal for a clean "Top 10" list.

### Resetting the Index

There are two common approaches:

**Approach 1 — Direct assignment:**
```python
df.index = range(1, 11)  # Set index to 1, 2, 3, ..., 10
```

**Approach 2 — Using `.reset_index()`:**
```python
df.reset_index(drop=True, inplace=True)  # Resets to 0, 1, 2, ...
```

The `drop=True` parameter discards the old index (instead of adding it as a new column).

### Why Start at 1?

In our case study, we use `range(1, 11)` to make the ranking more human-readable — the #1 country is at index 1, not index 0.

## 💻 Case Study Code

````python
# Get the top 10 countries
topTen = sorting.head(10)

# Reset the index to start from 1 (for a clean ranking)
topTen.index = range(1, 11)
````

**What just happened?**

1. `sorting.head(10)` — took the first 10 rows from our sorted DataFrame (the 10 countries with the highest obesity rates among those with population > 10M)
2. `topTen.index = range(1, 11)` — replaced the messy original index with a clean 1-to-10 ranking

# 🔗 Section 9: Putting It All Together — The Complete Data Management Pipeline

---

Now that you understand each individual concept, let's see the **entire solution** as a single, coherent pipeline. Notice how each step flows naturally into the next:

```
Import → Load → Rename → Merge → Select → Filter → Sort → Slice
```

````python
# ===== COMPLETE DATA MANAGEMENT SOLUTION =====

# Step 1: Import the library
import pandas as pd

# Step 2: Load the datasets
sample1 = pd.read_csv("../c2228.csv")   # Obesity data
sample2 = pd.read_csv("../c2119.csv")   # Population data

# Step 3: Rename columns for clarity and merge-readiness
sample1.rename(columns={"Pos": "Position", "Name": "Country", "Value": "Obesity Rate"}, inplace=True)
sample2.rename(columns={"Pos": "Position", "Name": "Country", "Value": "Population"}, inplace=True)

# Step 4: Merge the datasets on "Country"
joinedTable = pd.merge(sample1, sample2, how="left", on="Country")

# Step 5: Select only the columns we need
joinedTable1 = joinedTable[["Country", "Population", "Obesity Rate"]]

# Step 6: Filter — obesity > 20 AND population > 10 million
obesity_population = joinedTable1.loc[
    (joinedTable1["Population"] > 10000000) & (joinedTable1["Obesity Rate"] > 20)
]

# Step 7: Sort by Obesity Rate (highest first)
sorting = obesity_population.sort_values(by="Obesity Rate", ascending=False)

# Step 8: Get top 10 and set a clean index
topTen = sorting.head(10)
topTen.index = range(1, 11)

# Display the final result
topTen
````

## 📖 Concept: When and How to Use Bar Charts

### When to Use a Bar Chart

Bar charts are ideal when you want to **compare values across categories**. They answer questions like:
- Which country has the highest obesity rate?
- How do the top 10 countries compare?

### Vertical vs Horizontal

| Type | Function | Best When |
|------|----------|-----------|
| Vertical | `plt.bar()` | Category names are short |
| Horizontal | `plt.barh()` | Category names are long (like country names) |

When country names are long, **horizontal bars** prevent label overlap.

### `plt.figure(figsize=...)`

Controls the **size** of the chart:

```python
plt.figure(figsize=(width, height))  # in inches
```

### Adding Color

You can set bar colors using the `color` parameter:

```python
plt.barh(categories, values, color="steelblue")
```

Common color options: `'steelblue'`, `'coral'`, `'seagreen'`, `'goldenrod'`, `'tomato'`, `'#FF6B6B'` (hex codes), or use a **colormap** for gradient effects.

### `plt.tight_layout()`

This automatically adjusts spacing so that labels, titles, and axes don't overlap or get cut off. **Always add it** before `plt.show()`.

### Inverting the Y-Axis

By default, horizontal bar charts place the **first item at the bottom**. To show the #1 item at the **top** (more intuitive for rankings), use:

```python
plt.gca().invert_yaxis()
```

`gca()` means "get current axes" — it returns the active plot area.

## 💻 Case Study Code

````python
# --- Horizontal Bar Chart: Top 10 Countries by Obesity Rate ---

plt.figure(figsize=(10, 6))

plt.barh(topTen["Country"], topTen["Obesity Rate"], color="coral")

plt.title("Top 10 Countries by Obesity Rate (Population > 10M)", fontsize=14, fontweight="bold")
plt.xlabel("Obesity Rate (%)")
plt.ylabel("Country")

plt.gca().invert_yaxis()  # Highest at the top

plt.tight_layout()
plt.show()
````

**What just happened? — Line by line:**

| Line | What it does |
|------|--------------|
| `plt.figure(figsize=(10, 6))` | Creates a canvas 10 inches wide × 6 inches tall |
| `plt.barh(...)` | Draws horizontal bars — countries on Y-axis, obesity rates on X-axis |
| `color="coral"` | Sets all bars to a coral/salmon color |
| `plt.title(...)` | Adds a bold title with a larger font |
| `fontsize=14, fontweight="bold"` | Makes the title bigger and bolder |
| `plt.xlabel(...)` / `plt.ylabel(...)` | Labels the axes |
| `plt.gca().invert_yaxis()` | Flips the order so #1 (Egypt) is at the top |
| `plt.tight_layout()` | Prevents labels from being cut off |
| `plt.show()` | Renders the chart |

> 🔍 **Insight:** The chart makes it instantly clear that Egypt, USA, and Saudi Arabia lead in obesity rates among large countries — something that's harder to see in a raw table.

# Section 12: Customizing Charts — Colors, Annotations, and Style

---

## 📖 Concept: Making Charts Professional and Readable

A default chart gets the job done, but a **customized chart** tells a story. Here are the key customization tools:

### Data Labels / Annotations

Adding **values on top of bars** makes it easy to read exact numbers:

```python
for index, value in enumerate(data):
    plt.text(value, index, f" {value}", va='center')
```

| Parameter | What it does |
|-----------|--------------|
| `value` | X position of the text (end of the bar) |
| `index` | Y position (which bar) |
| `f" {value}"` | The text to display (the space adds padding) |
| `va='center'` | Vertically center the text on the bar |

### Color Gradients with Colormaps

Instead of one solid color, you can use a **colormap** to create a gradient:

```python
import numpy as np

colors = plt.cm.RdYlGn_r(np.linspace(0.2, 0.8, len(data)))
```

Popular colormaps:
- `plt.cm.RdYlGn_r` — Red → Yellow → Green (reversed)
- `plt.cm.Blues` — Light blue → Dark blue
- `plt.cm.viridis` — Purple → Green → Yellow
- `plt.cm.coolwarm` — Cool blue → Warm red

### Grid Lines

```python
plt.grid(axis='x', alpha=0.3)  # Subtle vertical gridlines
```

`alpha` controls transparency (0 = invisible, 1 = fully opaque).

### Removing Chart Borders (Spines)

```python
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
```

This gives the chart a cleaner, more modern look.

## 💻 Case Study Code

````python
import numpy as np

# --- Enhanced Horizontal Bar Chart with Customizations ---

plt.figure(figsize=(12, 7))

# Create a color gradient (red to green, reversed so high = red)
colors = plt.cm.RdYlGn_r(np.linspace(0.2, 0.8, len(topTen)))

# Draw the bars
bars = plt.barh(topTen["Country"], topTen["Obesity Rate"], color=colors, edgecolor="white")

# Add data labels at the end of each bar
for index, value in enumerate(topTen["Obesity Rate"]):
    plt.text(value + 0.3, index, f"{value}%", va='center', fontsize=11, fontweight='bold')

# Title and labels
plt.title("Top 10 Most Obese Countries\n(Among Nations with Population > 10 Million)",
          fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Obesity Rate (%)")

# Styling
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
````

**What just happened? — Key enhancements:**

| Enhancement | Code | Effect |
|-------------|------|--------|
| Color gradient | `plt.cm.RdYlGn_r(...)` | Bars go from red (high) to green (low) |
| White edges | `edgecolor="white"` | Separates the bars visually |
| Data labels | `plt.text(...)` loop | Shows exact % value next to each bar |
| Subtitle | `\n` in title | Creates a two-line title |
| Grid | `plt.grid(axis='x', alpha=0.3)` | Subtle reference lines |
| Clean borders | `spines[...].set_visible(False)` | Removes top and right borders |

This is the difference between a "homework chart" and a **presentation-ready visualization**.

# Section 13: Scatter Plots — Exploring Relationships

---

## 📖 Concept: When and How to Use Scatter Plots

### When to Use a Scatter Plot

Scatter plots show the **relationship between two numerical variables**. Each point represents one observation (in our case, one country). They answer questions like:

- Is there a relationship between population and obesity?
- Do bigger countries tend to have higher or lower obesity rates?

### `plt.scatter()` Syntax

```python
plt.scatter(x_data, y_data, s=size, c=color, alpha=transparency)
```

| Parameter | What it does |
|-----------|--------------|
| `x_data` | Values for the horizontal axis |
| `y_data` | Values for the vertical axis |
| `s` | Size of each dot (in points²) — can be a number or an array |
| `c` | Color of each dot — can be a single color or an array of values (for color mapping) |
| `alpha` | Transparency (0 = invisible, 1 = fully opaque) |
| `cmap` | The colormap to use when `c` is an array of numbers |
| `edgecolors` | Color of the dot border |

### Colorbars

When you color-code dots by a third variable, add a **colorbar** to explain what the colors mean:

```python
plt.colorbar(label="Obesity Rate (%)")
```

### Annotating Specific Points

You can label individual dots with country names:

```python
plt.annotate("USA", (x_position, y_position))
```

## 💻 Case Study Code

````python
# --- Scatter Plot: Population vs Obesity Rate ---

# Use the full filtered dataset (all countries with obesity > 20 and population > 10M)
plot_data = obesity_population.copy()

plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    plot_data["Population"],
    plot_data["Obesity Rate"],
    s=100,
    c=plot_data["Obesity Rate"],
    cmap="YlOrRd",
    alpha=0.8,
    edgecolors="gray"
)

# Add a colorbar
plt.colorbar(scatter, label="Obesity Rate (%)")

# Label each point with the country name
for _, row in plot_data.iterrows():
    plt.annotate(
        row["Country"],
        (row["Population"], row["Obesity Rate"]),
        fontsize=8,
        textcoords="offset points",
        xytext=(5, 5)
    )

plt.title("Population vs Obesity Rate\n(Countries with Pop > 10M & Obesity > 20%)",
          fontsize=14, fontweight="bold")
plt.xlabel("Population", fontsize=12)
plt.ylabel("Obesity Rate (%)", fontsize=12)
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

**What just happened? — Key components:**

| Component | Code | Effect |
|-----------|------|--------|
| Dot color mapping | `c=plot_data["Obesity Rate"]` | Colors each dot by its obesity rate |
| Colormap | `cmap="YlOrRd"` | Yellow (low) → Orange → Red (high) |
| Colorbar | `plt.colorbar(...)` | Shows the color legend |
| Labels | `plt.annotate(...)` in loop | Puts country names next to dots |
| `textcoords="offset points"` | — | Positions text relative to the dot |
| `xytext=(5, 5)` | — | Shifts the label 5 points right and up |

> 🔍 **Insight:** The scatter plot reveals that there's **no strong relationship** between population size and obesity rate — the USA is a large-population high-obesity outlier, while many high-obesity countries have relatively small populations.

# Section 14: Pie and Donut Charts — Showing Proportions

---

## 📖 Concept: When and How to Use Pie Charts

### When to Use a Pie Chart

Pie charts show **how parts make up a whole**. They answer questions like:
- What share of the total population does each top-10 country represent?
- How is the population distributed among the most obese nations?

### ⚠️ Important Limitations

Pie charts work best when:
- You have **fewer than 7–8 categories**
- You want to show **proportions of a total**
- The differences between slices are **significant**

They are **NOT ideal** when:
- You have many categories (too many thin slices)
- You need to compare precise values (bar charts are better)

### `plt.pie()` Syntax

```python
plt.pie(values, labels=names, autopct='%1.1f%%', startangle=90)
```

| Parameter | What it does |
|-----------|--------------|
| `values` | The numbers that determine slice sizes |
| `labels` | Text labels for each slice |
| `autopct` | Format for showing percentages inside slices |
| `startangle` | Rotation of the first slice (90 = start at 12 o'clock) |
| `explode` | A tuple that "pulls out" specific slices |
| `colors` | A list of colors for each slice |
| `wedgeprops` | A dict to style the wedges (e.g., `{'linewidth': 2, 'edgecolor': 'white'}`) |

### Making a Donut Chart

A donut chart is just a pie chart with a **white circle** in the centre:

```python
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)
```

## 💻 Case Study Code

````python
# --- Donut Chart: Population Distribution of Top 10 Obese Countries ---

plt.figure(figsize=(10, 10))

# Explode the largest slice slightly
explode = [0.05 if i == 0 else 0 for i in range(len(topTen))]

# Draw the pie chart
wedges, texts, autotexts = plt.pie(
    topTen["Population"],
    labels=topTen["Country"],
    autopct='%1.1f%%',
    startangle=90,
    explode=explode,
    wedgeprops={'linewidth': 2, 'edgecolor': 'white'}
)

# Style the percentage text
for autotext in autotexts:
    autotext.set_fontsize(9)
    autotext.set_fontweight('bold')

# Add a white circle in the centre to make it a donut
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(centre_circle)

# Add a title in the centre
plt.text(0, 0, 'Population\nShare', ha='center', va='center', fontsize=14, fontweight='bold')

plt.title("Population Distribution Among Top 10 Obese Countries",
          fontsize=14, fontweight="bold", pad=20)

plt.tight_layout()
plt.show()
````

**What just happened? — Key components:**

| Component | Code | Effect |
|-----------|------|--------|
| Explode | `[0.05 if i == 0 ...]` | Pulls the first (largest) slice out slightly |
| `autopct='%1.1f%%'` | — | Shows percentages with 1 decimal place |
| `wedgeprops={...}` | — | Adds white borders between slices |
| Centre circle | `plt.Circle(...)` | Transforms the pie into a donut |
| Centre text | `plt.text(0, 0, ...)` | Adds a label inside the donut hole |

# Section 15: Subplots — Multiple Charts in One Figure

---

## 📖 Concept: Arranging Multiple Charts Together

### Why Subplots?

Sometimes you want to show **related charts side by side** for comparison. Instead of creating separate figures, you can use **subplots** to place multiple charts in a single figure.

### `plt.subplots()` Syntax

```python
fig, axes = plt.subplots(nrows, ncols, figsize=(width, height))
```

| Parameter | What it does |
|-----------|--------------|
| `nrows` | Number of rows in the grid |
| `ncols` | Number of columns in the grid |
| `figsize` | Size of the entire figure |

This returns:
- `fig` — the overall figure object
- `axes` — an array of individual plot areas

### Accessing Individual Subplots

```python
# For a 1×2 grid (one row, two columns)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Draw on each subplot
ax1.bar(x, y)           # Left chart
ax2.scatter(x, y)       # Right chart

# Set titles for each subplot
ax1.set_title("Chart 1")
ax2.set_title("Chart 2")
```

### Key Difference from `plt.xxx()`

When using subplots, you call methods on the **axes object** (`ax.bar(...)`) instead of `plt.bar(...)`. The method names change slightly:

| `plt.xxx()` style | `ax.xxx()` style |
|-------------------|------------------|
| `plt.title()` | `ax.set_title()` |
| `plt.xlabel()` | `ax.set_xlabel()` |
| `plt.ylabel()` | `ax.set_ylabel()` |

## 💻 Case Study Code

````python
# --- Side-by-Side Comparison: Obesity Rate vs Population ---

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# --- LEFT CHART: Obesity Rate (horizontal bar) ---
colors1 = plt.cm.Reds(np.linspace(0.3, 0.9, len(topTen)))
ax1.barh(topTen["Country"], topTen["Obesity Rate"], color=colors1, edgecolor="white")
ax1.set_title("Obesity Rate (%)", fontsize=13, fontweight="bold")
ax1.set_xlabel("Obesity Rate (%)")
ax1.invert_yaxis()
ax1.grid(axis='x', alpha=0.3)
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

# Add data labels
for i, v in enumerate(topTen["Obesity Rate"]):
    ax1.text(v + 0.3, i, f"{v}%", va='center', fontsize=10)

# --- RIGHT CHART: Population (horizontal bar) ---
colors2 = plt.cm.Blues(np.linspace(0.3, 0.9, len(topTen)))
ax2.barh(topTen["Country"], topTen["Population"], color=colors2, edgecolor="white")
ax2.set_title("Population", fontsize=13, fontweight="bold")
ax2.set_xlabel("Population")
ax2.invert_yaxis()
ax2.grid(axis='x', alpha=0.3)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

# Add data labels
for i, v in enumerate(topTen["Population"]):
    ax2.text(v + 500000, i, f"{v:,.0f}", va='center', fontsize=9)

# Overall figure title
fig.suptitle("Top 10 Most Obese Countries (Pop > 10M) — Dual Comparison",
             fontsize=15, fontweight="bold", y=1.02)

plt.tight_layout()
plt.show()
````

**What just happened? — Key concepts:**

| Concept | Code | Effect |
|---------|------|--------|
| Create 2 subplots | `plt.subplots(1, 2, ...)` | 1 row, 2 columns |
| Unpack axes | `(ax1, ax2)` | Named variables for each subplot |
| Draw on specific subplot | `ax1.barh(...)` | Draws only on the left chart |
| Subplot titles | `ax1.set_title(...)` | Each chart gets its own title |
| Figure-level title | `fig.suptitle(...)` | One title above both charts |
| Different colormaps | `Reds` vs `Blues` | Visual distinction between metrics |

> 🔍 **Insight:** Placing the charts side-by-side reveals an interesting contrast — Egypt leads in obesity rate but the USA dwarfs all others in population size.

# Section 16: Building a Complete Dashboard

---

## 📖 Concept: Combining Everything into a Visual Story

A **dashboard** brings together multiple visualizations into a single view that tells a comprehensive story. In the real world, dashboards are used by:

- **Business analysts** to track KPIs
- **Data scientists** to present findings
- **Government agencies** to report public health data

### Design Principles

1. **Start with the big picture** — summary numbers or the most important chart first
2. **Use consistent colors** — same metric = same color throughout
3. **Don't overcrowd** — each chart should have enough space to breathe
4. **Label everything** — titles, axes, and annotations

### `gridspec` for Advanced Layouts

For non-uniform grids (e.g., one chart spanning two columns), use `gridspec_kw` or `matplotlib.gridspec`:

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
```

This creates a 2×2 grid with 4 subplot areas.

## 💻 Case Study Code — The Final Dashboard

````python
# ===== COMPLETE DASHBOARD: Population & Obesity Analysis =====

fig, axes = plt.subplots(2, 2, figsize=(18, 14))
fig.suptitle("🌍 Global Obesity & Population Dashboard\nTop 10 Countries (Population > 10M, Obesity Rate > 20%)",
             fontsize=16, fontweight='bold', y=1.02)

# --- Chart 1 (Top-Left): Horizontal Bar — Obesity Rates ---
ax1 = axes[0, 0]
colors_red = plt.cm.Reds(np.linspace(0.3, 0.85, len(topTen)))
ax1.barh(topTen["Country"], topTen["Obesity Rate"], color=colors_red, edgecolor="white")
for i, v in enumerate(topTen["Obesity Rate"]):
    ax1.text(v + 0.3, i, f"{v}%", va='center', fontsize=9, fontweight='bold')
ax1.set_title("🍔 Obesity Rate by Country", fontsize=12, fontweight="bold")
ax1.set_xlabel("Obesity Rate (%)")
ax1.invert_yaxis()
ax1.grid(axis='x', alpha=0.3)
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

# --- Chart 2 (Top-Right): Horizontal Bar — Population ---
ax2 = axes[0, 1]
colors_blue = plt.cm.Blues(np.linspace(0.3, 0.85, len(topTen)))
ax2.barh(topTen["Country"], topTen["Population"], color=colors_blue, edgecolor="white")
for i, v in enumerate(topTen["Population"]):
    ax2.text(v + 500000, i, f"{v:,.0f}", va='center', fontsize=8)
ax2.set_title("👥 Population by Country", fontsize=12, fontweight="bold")
ax2.set_xlabel("Population")
ax2.invert_yaxis()
ax2.grid(axis='x', alpha=0.3)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

# --- Chart 3 (Bottom-Left): Scatter Plot ---
ax3 = axes[1, 0]
scatter = ax3.scatter(
    topTen["Population"], topTen["Obesity Rate"],
    s=150, c=topTen["Obesity Rate"], cmap="YlOrRd",
    alpha=0.9, edgecolors="gray", zorder=5
)
for _, row in topTen.iterrows():
    ax3.annotate(row["Country"], (row["Population"], row["Obesity Rate"]),
                fontsize=8, textcoords="offset points", xytext=(5, 5))
ax3.set_title("📊 Population vs Obesity Rate", fontsize=12, fontweight="bold")
ax3.set_xlabel("Population")
ax3.set_ylabel("Obesity Rate (%)")
ax3.grid(alpha=0.3)
fig.colorbar(scatter, ax=ax3, label="Obesity Rate (%)")

# --- Chart 4 (Bottom-Right): Donut Chart ---
ax4 = axes[1, 1]
wedges, texts, autotexts = ax4.pie(
    topTen["Population"], labels=topTen["Country"],
    autopct='%1.1f%%', startangle=90,
    wedgeprops={'linewidth': 2, 'edgecolor': 'white'}
)
for autotext in autotexts:
    autotext.set_fontsize(8)
centre_circle = plt.Circle((0, 0), 0.65, fc='white')
ax4.add_artist(centre_circle)
ax4.text(0, 0, 'Population\nShare', ha='center', va='center', fontsize=11, fontweight='bold')
ax4.set_title("🥧 Population Distribution", fontsize=12, fontweight="bold")

plt.tight_layout()
plt.show()
````
